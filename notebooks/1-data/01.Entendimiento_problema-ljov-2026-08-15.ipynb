## ¿Cuál es el objetivo del problema?
Las enfermedades cardiovasculares constituyen una de las principales problemáticas de salud pública y son actualmente la principal causa de muerte en el mundo. Según la Organización Mundial de la Salud, en 2022 ocasionaron aproximadamente 19,8 millones de muertes, equivalentes al 32 % de todas las defunciones globales. Además de su impacto en la mortalidad y la calidad de vida, estas enfermedades representan una carga importante para los sistemas de salud debido a que pueden requerir seguimiento continuo, medicamentos, tecnologías diagnósticas, procedimientos especializados y atención hospitalaria.

Muchas enfermedades cardiovasculares pueden prevenirse o controlarse mediante la intervención sobre factores de riesgo modificables. Su identificación oportuna permite iniciar tempranamente acciones como orientación sobre hábitos saludables, actividad física, alimentación, abandono del tabaco, seguimiento médico y tratamiento farmacológico cuando sea necesario. Por esta razón, contar con herramientas que apoyen la identificación de personas con posibles problemas cardíacos puede contribuir a una atención más oportuna.

En este contexto, el objetivo del proyecto es desarrollar una prueba de concepto de aprendizaje automático que permita estimar si una persona presenta o no una enfermedad cardíaca a partir de sus características demográficas, signos y síntomas y resultados de exámenes clínicos, como edad, sexo, tipo de dolor torácico, presión arterial, colesterol, glucosa en ayunas, resultados del electrocardiograma y frecuencia cardíaca máxima.

La variable objetivo es `disease`, definida de la siguiente manera:

* `1`: presencia de enfermedad cardíaca.
* `0`: ausencia de enfermedad cardíaca.

La solución tendría un propósito académico y de apoyo para la identificación de posibles casos. Su resultado no debe interpretarse como un diagnóstico médico definitivo ni utilizarse para reemplazar la valoración de un profesional de la salud.


## ¿Cómo se usará la solución?
La solución podría utilizarse como una herramienta de apoyo para identificar personas cuyos datos demográficos, signos, síntomas y resultados de exámenes presenten patrones asociados con una enfermedad cardíaca.

El modelo recibiría las características de una persona y generaría una predicción sobre la posible presencia o ausencia de la condición. Esta información podría servir para priorizar una evaluación médica más detallada y apoyar la toma de decisiones de los profesionales de la salud.

Dentro del curso, la solución se desarrollará como una prueba de concepto para aplicar las etapas de preparación de datos, análisis exploratorio, feature engineering, entrenamiento, evaluación, interpretación y demostración de un modelo de Machine Learning.

El resultado de la prediccion no debe interpretarse como diagnóstico médico definitivo ni como reemplazo de la valoración clínica.

## ¿Cuáles son las soluciones actuales?
Actualmente, este tipo de problema se resuelve mediante la evaluación de un profesional de la salud, quien combina los antecedentes, síntomas, factores de riesgo, examen físico y resultados de pruebas como presión arterial, análisis de sangre, electrocardiograma, prueba de esfuerzo y otros estudios especializados.

## ¿Cómo se debe enmarcar este problema?

El problema se puede enmarcar como:

Aprendizaje supervisado, porque se dispone de ejemplos con una respuesta conocida.
Clasificación binaria, porque la variable objetivo tiene dos posibles resultados: 0 y 1.
Procesamiento fuera de línea o batch, porque se trabaja con un dataset histórico y estático.
Datos tabulares, porque cada fila representa una observación y cada columna una característica.
Predicción individual, porque posteriormente el modelo podría recibir los datos de un paciente y generar una clasificación.

## ¿Cómo se debe medir el desempeño de la solución?

El desempeño de la solución debería medirse principalmente mediante el **recall o sensibilidad de la clase positiva**, es decir, de los pacientes que realmente presentan una enfermedad cardíaca, qué proporción logra identificar correctamente el modelo.

Esta métrica se selecciona porque permite evaluar directamente la capacidad del modelo para reducir los falsos negativos, es decir, los pacientes que tienen la condición, pero son clasificados como si no la tuvieran.

El recall se complementará con:

Precision, para medir qué proporción de las predicciones positivas corresponde realmente a pacientes con la condición.
Especificidad, para evaluar la identificación correcta de pacientes sin la condición.
F1-score, para analizar el equilibrio entre recall y precision.
Matriz de confusión, para observar directamente los falsos positivos y falsos negativos.
ROC-AUC, para evaluar la capacidad general del modelo para diferenciar las dos clases.

La accuracy se podrá reportar como información complementaria, pero no será la métrica principal porque considera que los falsos positivos y falsos negativos tienen el mismo impacto.

En conclusión, la métrica principal propuesta es el **recall de la clase `disease = 1`**, debido a que el propósito de la solución sería apoyar la priorización de posibles pacientes cardíacos y reducir el riesgo de pasar por alto personas que necesiten una evaluación oportuna.

## ¿La medida de desempeño está alineada con el objetivo del problema?
Sí. La medida está alineada siempre que el modelo se utilice como una herramienta de priorización y apoyo, y no como un mecanismo autónomo para diagnosticar o iniciar tratamientos.

La selección de esta métrica se basa en el impacto de los diferentes resultados que podría producir el modelo:

| Resultado          | Interpretación                                                               | Posible impacto                                                                            |
| ------------------ | ---------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------ |
| Verdadero positivo | El paciente tiene la condición y el modelo lo identifica                     | Puede priorizarse correctamente para valoración médica y exámenes adicionales.             |
| Falso negativo     | El paciente tiene la condición, pero el modelo no la identifica              | Podría retrasarse la valoración, el diagnóstico y el inicio de intervenciones oportunas.   |
| Falso positivo     | El paciente no tiene la condición, pero el modelo lo clasifica como positivo | Podría recibir valoraciones o exámenes adicionales que finalmente descarten la enfermedad. |
| Verdadero negativo | El paciente no tiene la condición y el modelo lo clasifica correctamente     | Se evita utilizar recursos adicionales sin una indicación inicial.                         |

En este contexto, el error potencialmente más grave es el **falso negativo**, porque implicaría pasar por alto a una persona que podría beneficiarse de una valoración médica prioritaria, exámenes adicionales y acciones oportunas para evitar complicaciones. Por esta razón, se buscaría que el modelo tenga una sensibilidad alta, aunque esto implique aceptar cierta cantidad de falsos positivos.

Un falso positivo tendría inicialmente un impacto menor porque la salida del modelo no produciría un diagnóstico ni iniciaría automáticamente un tratamiento. Su consecuencia sería principalmente una evaluación clínica adicional, durante la cual podría descartarse la enfermedad. Además, las recomendaciones generales relacionadas con alimentación saludable, actividad física y control de factores de riesgo pueden beneficiar a la persona independientemente del resultado final.

Sin embargo, los falsos positivos tampoco deben considerarse inofensivos, ya que pueden generar costos, utilización innecesaria de recursos y, dependiendo del proceso de atención, pruebas diagnósticas invasivas. Por esta razón, el recall debe analizarse junto con la **precision**, que permitirá medir cuántas de las personas clasificadas como positivas realmente presentan la condición.

## ¿Cuál sería el desempeño mínimo necesario?

Como primera hipótesis del proyecto, se propone que el modelo alcance un recall mínimo de 80 % para la clase positiva, manteniendo al mismo tiempo una precision y especificidad aceptables.

Además, el modelo final deberá superar el desempeño del modelo baseline que se construirá en una actividad posterior.

El valor de 80 % es un objetivo académico inicial y no un estándar clínico. Este criterio deberá revisarse después de explorar los datos, construir el baseline y evaluar los resultados mediante validación cruzada.

## ¿Cómo se puede resolver el problema manualmente?

El problema podría resolverse inicialmente mediante con los factores de riesgo y los resultados clínicos disponibles en el conjunto de datos. Después de limpiar y validar los datos, se podrían definir teóricamente reglas como las siguientes:

1. **Factores demográficos (`age` y `sex`):** considerar como factor de riesgo que el paciente sea hombre y tenga 45 años o más, o que sea mujer y tenga 55 años o más. Estos rangos corresponden a edades a partir de las cuales aumenta el riesgo de enfermedad coronaria, aunque la edad por sí sola no sería suficiente para clasificar al paciente.

2. **Factores metabólicos y cardiovasculares:**

   * `rest_bp >= 140`: presión arterial sistólica elevada. La OMS utiliza 140/90 mmHg como referencia para hipertensión, aunque su diagnóstico requiere mediciones repetidas y en el dataset solamente se dispone de una medición de presión sistólica en reposo.
   * `chol >= 240`: colesterol total alto.
   * `fbs == 1`: glucosa en ayunas superior a 120 mg/dl, de acuerdo con la codificación propia del conjunto de datos.

3. **Síntomas y resultados funcionales:**

   * `chest_pain`: asignar una regla según el tipo de dolor torácico registrado.
   * `exang == 1`: presencia de angina inducida por el ejercicio.
   * `rest_ecg != "normal"`: existencia de alguna alteración en el electrocardiograma en reposo.
   * `old_peak >= 1`: depresión del segmento ST inducida por el ejercicio de al menos 1 mm, utilizada como posible indicador de isquemia durante una prueba de esfuerzo.

4. **Resultados de estudios complementarios:**

   * `ca >= 1`: presencia de uno o más vasos principales identificados mediante fluoroscopia, según la codificación del dataset.
   * `thal` igual a `fixed` o `reversable`: presencia de un defecto fijo o reversible en el estudio de talasemia.

Estas condiciones se podrían combinar mediante reglas anidadas. Por ejemplo:

* Si `ca >= 1` o `thal` presenta un defecto, clasificar como `disease = 1`.
* En caso contrario, si `exang == 1` y `old_peak >= 1`, clasificar como `disease = 1`.
* En caso contrario, si el paciente cumple el criterio de edad y sexo y, además, presenta dos o más factores entre presión elevada, colesterol alto, glucosa elevada o electrocardiograma anormal, clasificar como `disease = 1`.
* Si no se cumple ninguna de las combinaciones anteriores, clasificar como `disease = 0`.

Aunque este método sería interpretable y fácil de implementar inicialmente, al incorporar todas las variables y sus posibles interacciones se necesitaría una cantidad creciente de condiciones anidadas. Esto haría más difícil mantener las reglas y justificaría evaluar si un modelo de aprendizaje automático logra reconocer esas relaciones con mejor desempeño.



## Hacer un listado de los supuestos que hay hasta este momento.
Hasta este momento se plantean los siguientes supuestos:

- Cada fila representa a un paciente diferente y no varias consultas o mediciones de la misma persona. El dataset no contiene un identificador que permita comprobarlo directamente.
- La variable disease representa correctamente la presencia (1) o ausencia (0) de enfermedad cardiaca y fue construida a partir de un diagnóstico clínico confiable.
- Los valores diferentes de 0 y 1 encontrados en disease, así como los valores atípicos presentes en otras columnas, se interpretarán como errores de calidad y no como categorías adicionales con significado clínico.
- Las unidades y categorías descritas en el archivo informativo son correctas. Por ejemplo, rest_bp está expresada en mmHg, chol en mg/dl y old_peak representa la depresión del segmento ST.
- La población del dataset es suficientemente representativa de la población en la que se espera utilizar la solución.
- Las variables utilizadas para predecir están disponibles antes de conocer el diagnóstico. Este supuesto debe revisarse especialmente para ca y thal, porque corresponden a estudios especializados que posiblemente se realizan como parte del proceso diagnóstico.
- Si el objetivo es realizar una detección temprana con información demográfica, síntomas y exámenes básicos, podría ser necesario excluir ca y thal para evitar que el modelo dependa de información que no estaría disponible en el momento de uso.
- La solución se utilizará como herramienta de apoyo para identificar posibles casos y no para reemplazar el diagnóstico de un profesional de la salud.
- Los falsos negativos tienen un impacto mayor que los falsos positivos, porque implican pasar por alto pacientes que podrían tener una enfermedad cardiaca. Por lo tanto, inicialmente se asume que la sensibilidad o recall de la clase positiva tendrá mayor importancia.
- El costo de realizar una valoración o examen adicional a un falso positivo es menor que el costo de no detectar oportunamente a un verdadero paciente cardiaco.
- Las reglas manuales y los umbrales planteados son únicamente una línea base teórica y no constituyen un protocolo clínico validado.
- El rendimiento mínimo aceptable y el umbral final de clasificación seguramente deberá ajustarse, teniendo en cuenta la sensibilidad de realizar predicciones en el area de la salud, y compararse con la solución manual basada en reglas.

Estos supuestos deberán verificarse durante la exploración y el análisis de los datos.
